```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
from datasets import load_dataset

# 1. Load the dataset directly from Hugging Face
print("Downloading the dataset...")
dataset = load_dataset("bitext/Bitext-customer-support-llm-chatbot-training-dataset")

# 2. Convert the training split to a Pandas DataFrame for better exploration
df = dataset['train'].to_pandas()

# 3. View the total count and the first few rows
print(f"Total number of tickets: {len(df)}")
display(df.head())

# 4. Visualize the distribution of ticket categories (intents)
plt.figure(figsize=(10,6))
df['intent'].value_counts().head(15).plot(kind='barh', color='skyblue')
plt.title('Top 15 Customer Support Ticket Categories')
plt.gca().invert_yaxis()  # Invert Y-axis to show the highest frequency at the top
plt.show()
```
